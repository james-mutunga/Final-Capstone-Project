## 3. Tourism Demand Analysis

In this section, we analyse international visitor arrivals to understand how tourism demand has changed over the three-year period.

We begin by comparing annual arrivals across 2023, 2024 and 2025.

```python
# Calculate total international arrivals for each year
annual_arrivals = df_arrivals[
    ["Total_2023", "Total_2024", "Total_2025"]
].sum()

annual_arrivals
```

### 3.1 Calculating Annual Growth

We will calculate the percentage change in international visitor arrivals between consecutive years.

This will help us assess whether tourism demand is growing and whether its rate of growth is increasing or slowing down.

```python
# Calculate year-on-year growth rates
growth_2024 = (
    (annual_arrivals["Total_2024"] - annual_arrivals["Total_2023"])
    / annual_arrivals["Total_2023"]
) * 100

growth_2025 = (
    (annual_arrivals["Total_2025"] - annual_arrivals["Total_2024"])
    / annual_arrivals["Total_2024"]
) * 100

print(f"2024 growth: {growth_2024:.2f}%")
print(f"2025 growth: {growth_2025:.2f}%")
```

### 3.2 Measuring the Increase in Arrivals

We will calculate the absolute increase in international visitor arrivals between consecutive years.

This complements the percentage growth rates by showing how many additional visitors arrived in each year.

```python
# Calculate the absolute increase in arrivals
increase_2024 = (
    annual_arrivals["Total_2024"] - annual_arrivals["Total_2023"]
)

increase_2025 = (
    annual_arrivals["Total_2025"] - annual_arrivals["Total_2024"]
)

print(f"2024 increase: {increase_2024:,} visitors")
print(f"2025 increase: {increase_2025:,} visitors")
```

### 3.3 Analysing Monthly Arrivals

We will compare monthly international visitor arrivals across 2023, 2024 and 2025.

This will help us identify seasonal patterns, peak travel months and periods of lower visitor activity.

```python
# Compare monthly total arrivals across the three years
df_arrivals[
    ["Month", "Total_2023", "Total_2024", "Total_2025"]
]
```

### 3.4 Calculating Monthly Year-on-Year Changes

We will calculate the percentage change in international visitor arrivals for each month between 2023 and 2024, and between 2024 and 2025.

This will help us identify months with the strongest growth and months where visitor arrivals declined.

```python
# Calculate monthly percentage changes
df_arrivals["Growth_2024_pct"] = (
    (df_arrivals["Total_2024"] - df_arrivals["Total_2023"])
    / df_arrivals["Total_2023"] * 100
)

df_arrivals["Growth_2025_pct"] = (
    (df_arrivals["Total_2025"] - df_arrivals["Total_2024"])
    / df_arrivals["Total_2024"] * 100
)

# Display monthly growth rates
df_arrivals[
    ["Month", "Growth_2024_pct", "Growth_2025_pct"]
].round(2)
```

### 3.5 Identifying Strongest and Weakest Growth Months

We will identify the months with the highest and lowest year-on-year growth rates for 2024 and 2025.

This will help us pinpoint the months that contributed most positively to tourism demand growth and those that experienced the weakest performance.

```python
# Identify the months with the highest and lowest growth rates

# 2024
highest_2024 = df_arrivals.loc[df_arrivals["Growth_2024_pct"].idxmax()]
lowest_2024 = df_arrivals.loc[df_arrivals["Growth_2024_pct"].idxmin()]

# 2025
highest_2025 = df_arrivals.loc[df_arrivals["Growth_2025_pct"].idxmax()]
lowest_2025 = df_arrivals.loc[df_arrivals["Growth_2025_pct"].idxmin()]

print("2024")
print(f"Highest growth: {highest_2024['Month']} ({highest_2024['Growth_2024_pct']:.2f}%)")
print(f"Lowest growth: {lowest_2024['Month']} ({lowest_2024['Growth_2024_pct']:.2f}%)")

print("\n2025")
print(f"Highest growth: {highest_2025['Month']} ({highest_2025['Growth_2025_pct']:.2f}%)")
print(f"Lowest growth: {lowest_2025['Month']} ({lowest_2025['Growth_2025_pct']:.2f}%)")
```

### 3.6 Analysing Arrivals by Entry Point

International visitors enter Kenya through different points, including Jomo Kenyatta International Airport (JKIA), Moi International Airport (MIA) and other border points.

We will compare arrivals across these entry points for 2023, 2024 and 2025 to understand their respective contributions to international tourism.

```python
# Calculate total arrivals by entry point for each year

entry_points = pd.DataFrame({
    "2023": [
        df_arrivals["MIA_2023"].sum(),
        df_arrivals["JKIA_2023"].sum(),
        df_arrivals["Other_Borders_2023"].sum()
    ],
    "2024": [
        df_arrivals["MIA_2024"].sum(),
        df_arrivals["JKIA_2024"].sum(),
        df_arrivals["Other_Borders_2024"].sum()
    ],
    "2025": [
        df_arrivals["MIA_2025"].sum(),
        df_arrivals["JKIA_2025"].sum(),
        df_arrivals["Other_Borders_2025"].sum()
    ]
}, index=["MIA", "JKIA", "Other Borders"])

entry_points
```

### 3.7 Measuring Entry-Point Growth

We will calculate the absolute and percentage change in arrivals at each entry point between 2023 and 2025.

This will help us compare the growth of the different entry points over the two-year period.

```python
# Calculate the absolute change between 2023 and 2025
entry_points["Change"] = entry_points["2025"] - entry_points["2023"]

# Calculate percentage growth between 2023 and 2025
entry_points["Growth_pct"] = (
    (entry_points["2025"] - entry_points["2023"])
    / entry_points["2023"] * 100
)

# Display the results
entry_points.round(2)
```

### 3.8 Comparing Entry-Point Shares

We will calculate the percentage of total international visitor arrivals passing through each entry point for every year.

This will help us understand how the distribution of international arrivals across Kenya's entry points has changed over time.

```python
# Calculate each entry point's share of total arrivals
entry_point_shares = entry_points[["2023", "2024", "2025"]].div(
    annual_arrivals.values, axis=1
) * 100

entry_point_shares.round(2)
```

### 3.9 Measuring Changes in Entry-Point Share

We will calculate the change in each entry point's share of international arrivals between 2023 and 2025.

This will show how the distribution of arrivals has shifted over the two-year period.

```python
# Calculate the change in market share between 2023 and 2025
entry_point_shares["Change_pp"] = (
    entry_point_shares["2025"] - entry_point_shares["2023"]
)

entry_point_shares.round(2)
```

### 3.10 Validating Entry-Point Totals

We will sum arrivals across all three entry points for each year and compare the results with the annual international arrivals totals.

This ensures that our entry-point analysis accounts for all reported arrivals.

```python
# Sum arrivals across all entry points
entry_points[["2023", "2024", "2025"]].sum()
```

### 4.2 Key Accommodation Indicators

We extract occupied bed-nights, available bed-nights and occupancy rates for 2015–2025.

These indicators will help us compare accommodation demand with available capacity and assess how hotel utilisation has changed over time.

```python
accommodation_data = df_table_12_7_raw.iloc[[36, 37, 38], [0] + list(range(2, 13))].copy()

accommodation_data.columns = [
    "Measure", "2015", "2016", "2017", "2018", "2019",
    "2020", "2021", "2022", "2023", "2024", "2025"
]

df_accommodation = accommodation_data.set_index("Measure").T.reset_index()
df_accommodation.rename(columns={"index": "Year"}, inplace=True)

df_accommodation
```

### 4.3 Year-on-Year Changes

We calculate the annual percentage change in occupied and available bed-nights, alongside the percentage-point change in occupancy rate.

This helps us determine whether accommodation demand is growing faster or slower than available capacity.

```python
df_accommodation.columns = [
    "Year",
    "Occupied_Bed_Nights",
    "Available_Bed_Nights",
    "Occupancy_Rate"
]

df_accommodation["Year"] = df_accommodation["Year"].astype(int)

df_accommodation["Occupied_Growth_%"] = (
    df_accommodation["Occupied_Bed_Nights"].pct_change() * 100
)

df_accommodation["Available_Growth_%"] = (
    df_accommodation["Available_Bed_Nights"].pct_change() * 100
)

df_accommodation["Occupancy_Change_pp"] = (
    df_accommodation["Occupancy_Rate"].diff()
)

df_accommodation.round(2)
```
