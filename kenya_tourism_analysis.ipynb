## 3. Tourism Demand Analysis

In this section, we analyse international visitor arrivals to understand how tourism demand has changed over the three-year period.

We begin by comparing annual arrivals across 2023, 2024 and 2025.

```python
# Calculate total international arrivals for each year
annual_arrivals = df_arrivals[
    ["Total_2023", "Total_2024", "Total_2025"]
].sum()

annual_arrivals
```

### 3.1 Calculating Annual Growth

We will calculate the percentage change in international visitor arrivals between consecutive years.

This will help us assess whether tourism demand is growing and whether its rate of growth is increasing or slowing down.

```python
# Calculate year-on-year growth rates
growth_2024 = (
    (annual_arrivals["Total_2024"] - annual_arrivals["Total_2023"])
    / annual_arrivals["Total_2023"]
) * 100

growth_2025 = (
    (annual_arrivals["Total_2025"] - annual_arrivals["Total_2024"])
    / annual_arrivals["Total_2024"]
) * 100

print(f"2024 growth: {growth_2024:.2f}%")
print(f"2025 growth: {growth_2025:.2f}%")
```

### 3.2 Measuring the Increase in Arrivals

We will calculate the absolute increase in international visitor arrivals between consecutive years.

This complements the percentage growth rates by showing how many additional visitors arrived in each year.

```python
# Calculate the absolute increase in arrivals
increase_2024 = (
    annual_arrivals["Total_2024"] - annual_arrivals["Total_2023"]
)

increase_2025 = (
    annual_arrivals["Total_2025"] - annual_arrivals["Total_2024"]
)

print(f"2024 increase: {increase_2024:,} visitors")
print(f"2025 increase: {increase_2025:,} visitors")
```

### 3.3 Analysing Monthly Arrivals

We will compare monthly international visitor arrivals across 2023, 2024 and 2025.

This will help us identify seasonal patterns, peak travel months and periods of lower visitor activity.

```python
# Compare monthly total arrivals across the three years
df_arrivals[
    ["Month", "Total_2023", "Total_2024", "Total_2025"]
]
```

### 3.4 Calculating Monthly Year-on-Year Changes

We will calculate the percentage change in international visitor arrivals for each month between 2023 and 2024, and between 2024 and 2025.

This will help us identify months with the strongest growth and months where visitor arrivals declined.

```python
# Calculate monthly percentage changes
df_arrivals["Growth_2024_pct"] = (
    (df_arrivals["Total_2024"] - df_arrivals["Total_2023"])
    / df_arrivals["Total_2023"] * 100
)

df_arrivals["Growth_2025_pct"] = (
    (df_arrivals["Total_2025"] - df_arrivals["Total_2024"])
    / df_arrivals["Total_2024"] * 100
)

# Display monthly growth rates
df_arrivals[
    ["Month", "Growth_2024_pct", "Growth_2025_pct"]
].round(2)
```

### 3.5 Identifying Strongest and Weakest Growth Months

We will identify the months with the highest and lowest year-on-year growth rates for 2024 and 2025.

This will help us pinpoint the months that contributed most positively to tourism demand growth and those that experienced the weakest performance.

```python
# Identify the months with the highest and lowest growth rates

# 2024
highest_2024 = df_arrivals.loc[df_arrivals["Growth_2024_pct"].idxmax()]
lowest_2024 = df_arrivals.loc[df_arrivals["Growth_2024_pct"].idxmin()]

# 2025
highest_2025 = df_arrivals.loc[df_arrivals["Growth_2025_pct"].idxmax()]
lowest_2025 = df_arrivals.loc[df_arrivals["Growth_2025_pct"].idxmin()]

print("2024")
print(f"Highest growth: {highest_2024['Month']} ({highest_2024['Growth_2024_pct']:.2f}%)")
print(f"Lowest growth: {lowest_2024['Month']} ({lowest_2024['Growth_2024_pct']:.2f}%)")

print("\n2025")
print(f"Highest growth: {highest_2025['Month']} ({highest_2025['Growth_2025_pct']:.2f}%)")
print(f"Lowest growth: {lowest_2025['Month']} ({lowest_2025['Growth_2025_pct']:.2f}%)")
```

### 3.6 Analysing Arrivals by Entry Point

International visitors enter Kenya through different points, including Jomo Kenyatta International Airport (JKIA), Moi International Airport (MIA) and other border points.

We will compare arrivals across these entry points for 2023, 2024 and 2025 to understand their respective contributions to international tourism.

```python
# Calculate total arrivals by entry point for each year

entry_points = pd.DataFrame({
    "2023": [
        df_arrivals["MIA_2023"].sum(),
        df_arrivals["JKIA_2023"].sum(),
        df_arrivals["Other_Borders_2023"].sum()
    ],
    "2024": [
        df_arrivals["MIA_2024"].sum(),
        df_arrivals["JKIA_2024"].sum(),
        df_arrivals["Other_Borders_2024"].sum()
    ],
    "2025": [
        df_arrivals["MIA_2025"].sum(),
        df_arrivals["JKIA_2025"].sum(),
        df_arrivals["Other_Borders_2025"].sum()
    ]
}, index=["MIA", "JKIA", "Other Borders"])

entry_points
```

### 3.7 Measuring Entry-Point Growth

We will calculate the absolute and percentage change in arrivals at each entry point between 2023 and 2025.

This will help us compare the growth of the different entry points over the two-year period.

```python
# Calculate the absolute change between 2023 and 2025
entry_points["Change"] = entry_points["2025"] - entry_points["2023"]

# Calculate percentage growth between 2023 and 2025
entry_points["Growth_pct"] = (
    (entry_points["2025"] - entry_points["2023"])
    / entry_points["2023"] * 100
)

# Display the results
entry_points.round(2)
```

### 3.8 Comparing Entry-Point Shares

We will calculate the percentage of total international visitor arrivals passing through each entry point for every year.

This will help us understand how the distribution of international arrivals across Kenya's entry points has changed over time.

```python
# Calculate each entry point's share of total arrivals
entry_point_shares = entry_points[["2023", "2024", "2025"]].div(
    annual_arrivals.values, axis=1
) * 100

entry_point_shares.round(2)
```

### 3.9 Measuring Changes in Entry-Point Share

We will calculate the change in each entry point's share of international arrivals between 2023 and 2025.

This will show how the distribution of arrivals has shifted over the two-year period.

```python
# Calculate the change in market share between 2023 and 2025
entry_point_shares["Change_pp"] = (
    entry_point_shares["2025"] - entry_point_shares["2023"]
)

entry_point_shares.round(2)
```

### 3.10 Validating Entry-Point Totals

We will sum arrivals across all three entry points for each year and compare the results with the annual international arrivals totals.

This ensures that our entry-point analysis accounts for all reported arrivals.

```python
# Sum arrivals across all entry points
entry_points[["2023", "2024", "2025"]].sum()
```

### 4.2 Key Accommodation Indicators

We extract occupied bed-nights, available bed-nights and occupancy rates for 2015–2025.

These indicators will help us compare accommodation demand with available capacity and assess how hotel utilisation has changed over time.

```python
accommodation_data = df_table_12_7_raw.iloc[[36, 37, 38], [0] + list(range(2, 13))].copy()

accommodation_data.columns = [
    "Measure", "2015", "2016", "2017", "2018", "2019",
    "2020", "2021", "2022", "2023", "2024", "2025"
]

df_accommodation = accommodation_data.set_index("Measure").T.reset_index()
df_accommodation.rename(columns={"index": "Year"}, inplace=True)

df_accommodation
```

### 4.3 Year-on-Year Changes

We calculate the annual percentage change in occupied and available bed-nights, alongside the percentage-point change in occupancy rate.

This helps us determine whether accommodation demand is growing faster or slower than available capacity.

```python
df_accommodation.columns = [
    "Year",
    "Occupied_Bed_Nights",
    "Available_Bed_Nights",
    "Occupancy_Rate"
]

df_accommodation["Year"] = df_accommodation["Year"].astype(int)

df_accommodation["Occupied_Growth_%"] = (
    df_accommodation["Occupied_Bed_Nights"].pct_change() * 100
)

df_accommodation["Available_Growth_%"] = (
    df_accommodation["Available_Bed_Nights"].pct_change() * 100
)

df_accommodation["Occupancy_Change_pp"] = (
    df_accommodation["Occupancy_Rate"].diff()
)

df_accommodation.round(2)
```

### 4.4 Inspecting Table 12.8 — Hotel Bed-Nights by Zone

We will examine Table 12.8 to understand how hotel bed-night activity is distributed across Kenya's tourism zones from 2015 to 2025.

This will help us identify which zones record the highest accommodation activity and how bed-night patterns have changed over time.

```python
df_table_12_8_raw = pd.read_excel(
    file_path,
    sheet_name="Table 12.8",
    header=None
)

print(df_table_12_8_raw.shape)
display(df_table_12_8_raw.head(20))
```

#### 4.4.1 Extracting and Cleaning Hotel Bed-Nights by Zone

We will extract the tourism zones and their annual occupied bed-night figures from Table 12.8.

We will remove unnecessary blank columns, standardise the year and zone labels, and reshape the data into a clean format for comparison across zones and years.

```python
df_zones = df_table_12_8_raw.iloc[3:13, [0] + list(range(2, 13))].copy()

df_zones.columns = [
    "Zone", "2015", "2016", "2017", "2018",
    "2019", "2020", "2021", "2022", "2023",
    "2024", "2025"
]

df_zones["Zone"] = (
    df_zones["Zone"]
    .str.replace(r"\s*[…\.]+\s*$", "", regex=True)
    .str.strip()
)

df_zones
```

We will compare the number of occupied hotel bed-nights across the ten tourism zones in 2025.

This will help identify the zones with the highest accommodation activity and show how bed-night activity is distributed geographically.

```python
zone_2025 = (
    df_zones[["Zone", "2025"]]
    .sort_values(by="2025", ascending=False)
    .reset_index(drop=True)
)

zone_2025
```

#### 4.4.3 Analysing Changes in Bed-Nights Across Zones

We will compare occupied bed-nights across tourism zones between 2015 and 2025.

This will help identify which zones have experienced the strongest growth or decline over the period and how accommodation activity has shifted geographically.

```python
zone_growth = df_zones[["Zone", "2015", "2025"]].copy()

zone_growth["Change"] = zone_growth["2025"] - zone_growth["2015"]

zone_growth["Growth_%"] = (
    (zone_growth["2025"] - zone_growth["2015"])
    / zone_growth["2015"] * 100
)

zone_growth.sort_values(
    by="Growth_%",
    ascending=False
).round(2).reset_index(drop=True)
```

#### 4.4.4 Assessing the Impact of 2020 and Subsequent Recovery

We will compare occupied hotel bed-nights in 2019, 2020 and 2025 across tourism zones.

This will help identify the extent of the decline in 2020 and assess how far each zone had recovered by 2025 compared with its pre-2020 level.

```python
zone_recovery = df_zones[["Zone", "2019", "2020", "2025"]].copy()

zone_recovery["Change_2019_2020_%"] = (
    (zone_recovery["2020"] - zone_recovery["2019"])
    / zone_recovery["2019"] * 100
)

zone_recovery["Change_2019_2025_%"] = (
    (zone_recovery["2025"] - zone_recovery["2019"])
    / zone_recovery["2019"] * 100
)

zone_recovery.round(2)
```

#### 4.4.5 Share of Occupied Bed-Nights by Zone in 2025

We will calculate each tourism zone's percentage share of total occupied bed-nights across the ten zones in 2025. This will show how accommodation activity is distributed geographically and how concentrated it is in the leading zones.

```python
zone_share_2025 = df_zones[["Zone", "2025"]].copy()

zone_share_2025["Share_%"] = (
    zone_share_2025["2025"]
    / zone_share_2025["2025"].sum()
    * 100
)

zone_share_2025 = (
    zone_share_2025
    .sort_values(by="Share_%", ascending=False)
    .round(2)
    .reset_index(drop=True)
)

zone_share_2025
```

## 4.5 Key Findings: Accommodation Performance

### Hotel Bed-Night Performance
- Occupied bed-nights increased from **5.88 million in 2015 to 11.56 million in 2025**.
- **2025 recorded the highest occupied bed-night figure** in the period analysed.

### Hotel Occupancy Rates
- Occupancy increased from **29.1% in 2015 to 33.8% in 2019**.
- It fell to **17.8% in 2020**, before recovering to **27.8% in 2025**.
- Despite the recovery, the 2025 occupancy rate remained below the pre-2020 level.

### Available Capacity
- In 2025, occupied bed-nights increased by **12.61%**, while available bed-nights increased by **17.11%**.
- Available capacity therefore grew faster than usage, contributing to the decline in occupancy from **28.9% in 2024 to 27.8% in 2025**.

### Accommodation Activity by Zone
- **Coastal Beach recorded the highest activity**, with 4.94 million occupied bed-nights in 2025.

### Growth Across Tourism Zones

- Coastal Beach recorded the largest absolute increase, adding **2.83 million occupied bed-nights** over the same period.
- Coastal Other recorded the largest decline (**-34.56%**).

### Recovery Across Tourism Zones
- **Seven of the ten zones exceeded their 2019 occupied bed-night levels by 2025**.
- Coastal Other, Nairobi Other, Western and Northern remained below their 2019 levels.
- The impact of the 2020 decline and subsequent recovery varied across zones.

### Overall Takeaway

### 6.1 Table 12.12: Indicators of Conference Tourism

This section examines Kenya's conference tourism performance from 2015 to 2025.

We will explore local and international conferences, delegate numbers, delegate-days and conference capacity utilisation.

```python
df_table_12_12_raw = pd.read_excel(
    file_path,
    sheet_name="Table 12.12",
    header=None
)

print(df_table_12_12_raw.shape)
display(df_table_12_12_raw.head(15))
```

### 6.1.1 Data Extraction and Cleaning

The source table presents conference tourism indicators across two categories: local and international.

The data will be reorganised into a structured format to support year-on-year comparisons and trend analysis.

```python
df_conference = df_table_12_12_raw.iloc[3:9, :].copy()

years = [str(year) for year in range(2015, 2026)]

columns = ["Indicator"]
for year in years:
    columns.extend([f"{year}_Local", f"{year}_International"])

columns = ["Indicator"] + columns[1:]

df_conference.columns = columns

df_conference["Indicator"] = (
    df_conference["Indicator"]
    .str.replace(r"\.+", "", regex=True)
    .str.strip()
)

df_conference = df_conference.set_index("Indicator").T
df_conference.index.name = "Year"

display(df_conference)
```

### 6.1.2 Preparing the Data for Analysis

Separate the year and conference category into their own columns.

This will make it easier to compare local and international conference tourism over time.

```python
df_conference = df_conference.reset_index()

df_conference[["Year", "Category"]] = (
    df_conference["Year"]
    .str.split("_", expand=True)
)

df_conference["Year"] = df_conference["Year"].astype(int)

display(df_conference)
```

### 6.1.3 Renaming Columns

Shorten the indicator names into clear column labels. This will make the DataFrame easier to work with during analysis.

```python
df_conference = df_conference.rename(columns={
    "Number of conferences …………………………………………": "Conferences",
    "Number of delegates  …………………………………………": "Delegates",
    "Number of delegate days ………………………………………………": "Delegate_Days",
    "Number of delegate days available ………………………………………………": "Available_Delegate_Days",
    "Percentage Occupancy …………………………………………………………": "Occupancy_Percent",
    "Conference Capacity Utilization (Percentage)": "Capacity_Utilization"
})

display(df_conference)
```

### 6.1.4 Data Quality Check

Check the data types and missing values before analysis.

This helps identify incomplete fields and confirms that the numerical indicators can be used in calculations.

```python
df_conference.info()

display(df_conference.isnull().sum())
```

### 6.1.5 Data Type Conversion

Convert the conference indicators from object to numeric data types.

This ensures the values can be used reliably in calculations, comparisons and visualisations.

```python
numeric_columns = [
    "Conferences",
    "Delegates",
    "Delegate_Days",
    "Available_Delegate_Days",
    "Occupancy_Percent",
    "Capacity_Utilization"
]

df_conference[numeric_columns] = df_conference[numeric_columns].apply(
    pd.to_numeric,
    errors="coerce"
)

df_conference.info()
```

### 6.2 Conference Tourism Performance

Compare local and international conference activity using the number of conferences and delegates recorded each year.

This will help identify which category contributes most to conference tourism and how participation has changed over time.

### 6.2.1 Year-on-Year Growth

Calculate the annual percentage change in the number of conferences and delegates for both local and international categories.

This will help identify periods of growth, decline and recovery.

```python
df_conference_growth = df_conference.sort_values(
    ["Category", "Year"]
).copy()

growth_columns = [
    "Conferences",
    "Delegates",
    "Delegate_Days"
]

for column in growth_columns:
    df_conference_growth[f"{column}_Growth_%"] = (
        df_conference_growth
        .groupby("Category")[column]
        .pct_change() * 100
    )

display(
    df_conference_growth[
        ["Year", "Category"] +
        [f"{column}_Growth_%" for column in growth_columns]
    ].round(2)
)
```

Examine local and international occupancy rates alongside overall conference capacity utilisation.
